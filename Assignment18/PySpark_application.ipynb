```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
```

```python
spark = SparkSession.builder.appName("Partition_assignment").getOrCreate()
```

```python
df = spark.range(0, 5000000)
```

```python
print("Before partitions:", df.rdd.getNumPartitions())
```

```python
df.write.mode("overwrite").csv("output/BeforePart", header=True)
```

```python
df_repartitioned = df.repartition(12)
```

```python
print("After partitions:", df_repartitioned.rdd.getNumPartitions())
```

```python
df_repartitioned.write.mode("overwrite").csv("output/afterPart", header=True)
```

```python
df_coal = df_repartitioned.coalesce(3)
print("After coalesced:", df_coal.rdd.getNumPartitions())
```

```python
df_coal.write.mode("overwrite").csv("output/afterCoalesce", header=True)
```

```python
spark.stop()
```
