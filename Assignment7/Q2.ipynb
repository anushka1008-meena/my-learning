```python
# 2)Create two DataFrames, df1 and df2, with a common column (e.g., 'ID'). 

import pandas as pd

df1 = pd.DataFrame({
    'ID': [1,2,3,4],
    'Name': ['A1','B1','C1','D1'],
    'subject': ['sub1','sub2','sub4','sub6'],
})

df2 = pd.DataFrame({
    'ID': [3,2,7,9],
    'Name': ['B1','C1','P1','Q1'],
    'subject': ['sub1','sub2','sub3','sub4'],
})

# (i) Perform an inner merge on this common column and display the resulting DataFrame.
a = df1.merge(df2, on = 'ID', how = 'inner')
print(a)
```

```python
# (ii) Perform a left join of df1 and df2 on the 'ID' column.

import pandas as pd

df1 = pd.DataFrame({
    'ID': [1,2,3,4],
    'Name': ['A1','B1','C1','D1'],
    'subject': ['sub1','sub2','sub4','sub6'],
})

df2 = pd.DataFrame({
    'ID': [3,2,7,9],
    'Name': ['B1','C1','P1','Q1'],
    'subject': ['sub1','sub2','sub3','sub4'],
})

print('\n left join of df1 and df2')
b = df1.merge(df2, on = ['ID'], how = 'left')
print(b)

# Explain how missing values are handled in the resulting DataFrame.

print('\n Explanation')
print('Left join = keeps all value from left (i.e. -> df1)')
print('if matching ID is NOT found in df2 -> then pandas fill missing values with NaN')

# Right Join and Index-Based Join

print('\n Right join')
c = df1.merge(df2, on='ID', how='right')
print(c)

print('\n Index-based join')
d = df1.join(df2, lsuffix='_df1', rsuffix='_df2')
print(d)
```

```python
# (iii) Perform a right join using pd.merge() on a common column  & Merging with Multiple Keys.
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1,2,3,4],
    'Name': ['A1','B1','C1','D1'],
    'subject': ['sub1','sub2','sub4','sub6'],
})

df2 = pd.DataFrame({
    'ID': [3,2,7,9],
    'Name': ['B1','C1','P1','Q1'],
    'subject': ['sub1','sub2','sub3','sub4'],
})

print('\n Right join using pd.merge()')
r_join = pd.merge(df1, df2, on = 'ID', how = 'right')
print(r_join)


#  Merging with Multiple Keys. 
print('\n merginf with multiple keys')
a = pd.merge(df1, df2, on=['ID','subject'], how='inner')
print(a)



# perform a join using df.join() based on the index. Compare the results.

print('\n index based join')
df1 = pd.DataFrame({
    'Name': ['A1','B1','C1','D1']
}, index=[1,2,3,4])

df2 = pd.DataFrame({
    'Marks': [80,90,70,60]
}, index=[3,2,7,9])

e = df1.join(df2, how='right')
print(e)

print('\n Comparision')
print('merge() = join dataframes using a common column like (ID)')
print('join() = joins dataframes using index -> here we gives index to both df1 & df2' )


```
