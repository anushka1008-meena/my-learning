```python
# 3) Create three DataFrames. 

import pandas as pd

one = pd.DataFrame({
    'ID': [1,2,3],
    'name': ['A','B','C']
})

two = pd.DataFrame({
    'ID': [5,4,3],
    'name': ['H','L','A']
})

three = pd.DataFrame({
    'ID': [7,4,3],
    'name': ['K','B','C']
})



# (i) Vertically concatenate two of them using pd.concat()
res = pd.concat([one,two], axis = 0, keys = ['x','y'])
print(res)



# (ii) then merge the resulting DataFrame with the third DataFrame on a common key.
a = pd.merge(res,three, on = 'ID')
print('\n merging resulting df with 3rd df')
print(a)



# (iii) understand join() vs. merge().

print('\n join() vs. merge()')
print(''' 
    merge() 
    (i) used to combine DataFrames using common columns , Uses -> pd.merge()
    (ii) similar to SQL joins
    (iii) More flexible and commonly used.
    
    join()
    (i) Used mainly for joining DataFrames using indexes.
    (ii) Simpler syntax
    (iii) Uses -> df.join()
''')



# (iv) Also Explain the primary differences between df.join() and pd.merge()

print('\n df.join() v/s pd.merge()')
print('''
    df.join()
    (i) Mainly used for index based joining
    (ii) simpler syntax 
    (iii) Called using dataframe object

    pd.merge()
    (i) Mainly used for column base joining
    (ii) more flexible & powerful
    (iii) Called using pandas function
''')
```
