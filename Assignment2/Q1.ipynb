```python
# Dictionary

d = { }
print(type(d))

marks = {
    'anu' : 79,
    'priya' : 69,
    'jiya' : 80
}

print(marks,type(marks))
print(len(marks))
print(marks['priya'])

print(marks.items())
print(marks.keys())
print(marks.values())
marks['priya'] = 98
print(marks)
print(marks.get('jiya'))

marks.update({'anu': 89, 4: 'Hello'})
print(marks)

a = marks.pop('priya')
print(marks)

del marks[4]
print(marks)

for x in marks:
    print(x)        # print keys
```

```python
# Tuple

t = (1,2,3,2,'box')
print(type(t))

t = (1)            # Here only single value is present that's why it's type = int
print(type(t))  

t = (1,)
print(type(t))    # noe this consider as tuple

t = (1,2,3,2,2,'box')
a = t.count(2)            # it counts how many time 2 repeates
print(a)

# access partcular index
i = t.index('box')
print(i)

# check element exists or not
print(4 in t)
print(1 in t)

# to repeat tuple
b = t * 2
print(b)
```

```python
# Sets

s = {1,2,7,7,8,'anu'}     # sets are unordered & removes duplicate values
print(s)
print(len(s))

e = set()     # it is an empty set
print(type(e))

s.add('z')
s.remove(2)
print(s)

s.pop()
print(s)

s.clear()
print(s)     # gives empty set

s1 = {1,2,3,4}
s2 = {4,5,6}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print({5}.issubset(s2))             # subset = each element present in set
print({2,3,4}.issuperset(s1))       # superset = whole set
print(s2.pop())
print(s1.clear())
```
