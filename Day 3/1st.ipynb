```python
# List

l = []
print(l,type(l))

l1 = [10,20,30]
print(l1)

l2 = ['A',40,'B','C']
print(l2[-1])
print(l2[0:3])
print(l2[0::3])
print(l2[4::-1])
c = len(l2)
print(c)

print() 

for i in range(3):
    print(l2[i])
    print(l2[i] * 2)

print()

for x in l1:
    print(x)

print()
print(sum(l1))

# sum by using for loop
s = 0
for x in l1:
    s += x
print(s)

# find max element using for loop
max = l1[0]
for x in l1:
    if x > max:
        max = x
print(max)
```

```python
# Dynamic list

str = 'This is python'
l = str.split() # if it got space then it split it
print(l)

l1 = ['a','b',3,4]
l1.append(l)
print(l1)

# l1.remove('b')
# print(l1)

print(l1[4][0]) # 4th index ka 0 element print hoga

# append using for loop
for i in range(1,3):
    l1.append(i)
    print(l1)
```

```python
# Dictionary

d = {
    'name' : 'python',
    '2' : 'java',
    '3' : 100,
    '4' : 200,
    # nested dictionary
    'd2' : {                   # ? 
        'pname' : 'abc',
        'mrp' : 400 
    }
}
print(d, type(d))

# dictionary can't access directly by indexing
print(d['name'])
```

```python
d = {}
d[0] = 'A'   # 0 is not an index it is just like an variable which stores the value
d['name'] = 'java'
d['alpha'] = 'python'
print(d.keys())
print(d)

del(d[0])
print(d)
print(d.update({
    1: 'C++',
    2: 'Oops',
    3: 'Dsa'
    }))
print(d)

# del = key & value both are removed
```
