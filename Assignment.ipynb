```python
# Q2 = string

a = 'hello '
b = 'PYTHON programming'
c = a+b
print(c)
print(a.upper())
print(b.lower())
print(b.title())
print(b.swapcase())           # it swaps upper to lower & vice-versa
print(a.capitalize())
print(b.casefold())          # covert whole text in lowercase
print(c.center(40))
print(b.count('m'))
print(b.endswith('ing'))
print(b.find('N'))
print(a.isalnum())        # checks alphabet & numbers -> if space present = gives false
print(a.isdigit())
print(b.isnumeric())
print(a.isspace())       # If in string only space is present then it is true[like = " "]
print(a.replace('hello','hii'))
```

```python
# Q5

name = input('enter your name: ')
Class = input('enter your class:')

marks1 = int(input('m1'))
marks2 = int(input('m2'))
marks3 = int(input('m3'))
marks4 = int(input('m4'))
marks5 = int(input('m5'))

total_marks = marks1 + marks2 + marks3 + marks4 + marks5

total_percentage = (100*total_marks)/500

if(total_percentage>=60):
    grade = 'A'

elif(total_percentage>=50 and total_percentage<60):
    grade = 'B'

elif(total_percentage>=40 and total_percentage<50):
    grade = 'C'

elif(total_percentage>=33 and total_percentage<40):
    grade = 'D'

else:
    grade = 'FAIL'

print('name =',name)
print('Class =',Class)
print('total_marks =',total_marks)
print('total_percentage =',total_percentage)
print('grade =',grade)
```
