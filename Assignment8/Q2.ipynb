```python
# 2) Explore more datetime function and uses in pandas 

import pandas as pd

print(pd.Timestamp.now()) # it shows current date & time
print()

d = pd.to_datetime('2026-05-27')
print(d)
print()

print(d.year)
print(d.month)
print(d.day)

print()
dates = pd.date_range(start='2026-05-01', periods=4) # here we gives range of date as periods
print(dates)

print()
next_date = d + pd.Timedelta(days=7)  # it adds the no. of days in orignal date & then print that date
print(next_date)

# difference between dates
d1 = pd.to_datetime('2026-05-01')
d2 = pd.to_datetime('2026-05-10')
print(d2 - d1)
```
