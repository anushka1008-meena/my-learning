```python
# 3) Import an meaningful csv file for data analysis and perform data cleaning, and analysis for meaningful output

import pandas as pd

df = pd.read_csv("tips.csv")

print("First 5 rows")
print(df.head())       # it shows 5 rows of the dataset


print("\nDataset Info")
print(df.info())        # it gives information about dataset

# check missing values
print("\nMissing Values")         # in output if 0 occurs it means that no missing data is present in any column
print(df.isnull().sum())

df = df.drop_duplicates()    # it removes duplicate rows

# by using aggregate functions 
print("\nAverage tip")
print(df['tip'].mean())


print("\nMaximum bill")
print(df['total_bill'].max())


print("\nMinimum bill")
print(df['total_bill'].min())


print("\nGender count")
print(df['sex'].value_counts())  # it counts total males & females



# average total bill according to gender
print("\nAverage bill according to gender")
print(df.groupby('sex')['total_bill'].mean())



# highest tip
print("\nHighest tip")
print(df['tip'].max())
```
