```python
# 1) Explore more regex patterns Eg. The regex pattern used to validate email addresses, mobile no, string, and more 
 
# regex symbols meaning
# 1) ^ = start of string    2) $ = end of string     3) [] = range of characters   4) + = one or more        5) * = zero or more
# 6) {n} = exact no.        7) {n,} = minimum no.    8) \d = digits                9) \w = word characters   10) . = any character   

import re

# Email validation
email = "@gmail.com"

pattern = r'^[a-zA-Z0-9._]+@[a-z]+\.[a-z]{2,}$'

if re.match(pattern, email):
    print("Valid Email.....!!")
else:
    print("Invalid Email....!!")
```

```python
# mobile no. validation
import re

mobile = '9070003210'

pattern = r'^[0-9]{10}$' # it select range from 0 to 9 & with exact 10 no.'s

if re.match(pattern, mobile):
    print("Valid number")
else:
    print("Invalid number...!")
```

```python
# string validation
import re

name = 'Anushka'
# name = 'anu123'   -> it gives error because this takes only alphabets

pattern = r'^[A-Za-z]+$'

if re.match(pattern, name):
    print("Valid string")
else:
    print("Invalid string")
```

```python
# number validation

import re

num = '123a5'
# num = '12345'   -> it is valid 

pattern = r'^[0-9]+$'

if re.match(pattern, num):
    print("Only numbers")
else:
    print("Invalid....!")
```

```python
# password validation

import re

# password = 'abc'          ->  it includes only characters so it becomes invalid password
password = '123abc#'

pattern = r'^[A-Za-z0-9@#]{6,}$'

if re.match(pattern, password):
    print("Valid password")
else:
    print("Invalid password.......!!")
```
