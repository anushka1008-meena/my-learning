```python
# 1) Practise Pandas Series

# Create a Pandas Series from Dictionary 
import pandas as pd

data = {
    'Name':['anu','disha','divya'],
    'age':[19,20,18],
    'marks':[96,86,87]
}
df = pd.Series(data)
print(df) 

# ------------------------------------------------
# Create a Pandas Series from Lists 
print()
info = [10,20,30,40]
df = pd.Series(info)
print(df)

# ------------------------------------------------------------
# Access the elements of a Series in Pandas
print()
print(df[0])
print(df[2])
```
