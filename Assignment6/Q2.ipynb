```python
# 2) DataFrames

# Make a Pandas DataFrame with a two-dimensional Python list /  Create Pandas dataframe using list of lists = both questions are similar
import pandas as pd

data = [
    ['anu',19,90],
    ['disha',20,86],
    ['divya',18,87]
]
df = pd.DataFrame(data, columns = ['Name','age','marks'])
print(df)



# --------------------------------------------------------
# Create DataFrame from Python dict 
print()      # It is used for giving spaces b/w outputs
data = {
    'Name':['anu','disha','divya'],
    'age':[19,20,18],
    'marks':[96,86,87]
}
df = pd.DataFrame(data)
print(df)


# --------------------------------------------------------
# Create a Pandas dataframe using list of tuples 
print()
data = [
    ('anu',19,90),
    ('disha',20,86),
    ('divya',18,87)
]
df = pd.DataFrame(data, columns = ['Name','age','marks'])
print(df)


# ------------------------------------------------------------------
# Create a Pandas DataFrame from List of Dicts 
print()
data = [
    {'Name':'anu' , 'age':19 , 'marks':96},
    {'Name':'disha' , 'age':20 , 'marks':86},
    {'Name':'divya' , 'age':18 , 'marks':87}  
]
df = pd.DataFrame(data)
print(df)

```
