```python
# 3)Data iteration

import pandas as pd
data = {
    'Name': ['kirti','riya','kavya','shree'],
    'age': [19,20,21,18],
    'Marks': [90,87,68,75]
}
df = pd.DataFrame(data)
print(df)

# Different ways to iterate over rows in Pandas Dataframe

print("\nUsing iterrows()")
for i, row in df.iterrows():
    print(i, row['Name'], row['Marks'])

print("\nUsing itertuples()")
for row in df.itertuples():
    print(row.Name, row.age, row.Marks)




# ---------------------------------------------------
# Selecting rows in pandas DataFrame based on conditions
 
print("\nRows where Marks > 85")
print(df[df['Marks'] > 85])




# --------------------------------------------------
# Select any row from a Dataframe using iloc[]

print("\nSelect row using iloc[]")
print(df.iloc[1])         # here we use single bracket & single column only so its o/p is in the form of Series




# ---------------------------------------------
# Limited rows selection with given column

print("\nSelect only Name and Marks columns")
print(df[['Name','Marks']].head(2))




# --------------------------------------------------------------
# Drop rows from the dataframe based on certain condition applied on a column 
print("\nDrop rows where Marks > 80")

new = df[df['Marks'] < 80]
print(new)




# --------------------------------------------------------
# Insert row at given position in Pandas Dataframe 
new_row = pd.DataFrame({
    'Name':['sonal'],
    'age':[17],
    'Marks':[91]
})

# here we insert new data at index 2
df1 = pd.concat([df.iloc[:2], new_row, df.iloc[2:]]).reset_index(drop=True)

print("\nAfter inserting new row")
print(df1)




# -----------------------------------------------------------
# Create a list from rows in Pandas dataframe 
print()

row_list = df.values.tolist()
print(row_list)
```
