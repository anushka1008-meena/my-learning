```python
#  6) Solve the following equation using linalg() 
#     and inverse Matrix method x - 2y + 3z = 9 -x + 3y - z = -6 2x - 5y + 5z = 17 7) 
#     Using "Customizing Matplotlib Visualizations" discussed in Numpy session4 notes plot graph to compare your at least 2 semester result


#   x−2y+3z=9
#   −x+3y−z=−6
#   2x−5y+5z=17

import numpy as np

a = np.array([
    [1, -2, 3],
    [-1, 3, -1],
    [2, -5, 5]
])

b = np.array([9, -6 , 17])

res = np.linalg.solve(a,b)
print(res)

```

```python
# inverse matrix method

a = np.array([
    [1, -2, 3],
    [-1, 3, -1],
    [2, -5, 5]
])

b = np.array([9, -6 , 17])

a_inv = np.linalg.inv(a)
print(a_inv)

print()
res = np.dot(a_inv, b)    # here matrix multiplication occurs
print(res)
```
